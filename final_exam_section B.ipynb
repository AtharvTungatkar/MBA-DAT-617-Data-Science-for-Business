```python
# Question 18
import pandas as pd

pet_supplies = pd.read_csv("pet_supplies.csv")

# Count reviews missing review text but containing rating
missing_review_with_rating = pet_supplies[
    pet_supplies["review"].isna() & pet_supplies["rating"].notna()
]

print(len(missing_review_with_rating))

# Remove reviews missing review text before proceeding
pet_supplies = pet_supplies.dropna(subset=["review"])

print(pet_supplies.shape)
```

```python
# Question 21
from nltk.tokenize import word_tokenize

pet_supplies["num_words"] = pet_supplies["review"].apply(
    lambda x: len(word_tokenize(x, preserve_line=True))
)

answer = pet_supplies["num_words"].mean()

print(answer)
```

```python
# Question 23

import nltk
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger_eng')

pet_supplies["proper_noun_count"] = pet_supplies["review"].apply(
    lambda x: sum(
        1 for word, tag in pos_tag(word_tokenize(x, preserve_line=True))
        if tag in ["NNP", "NNPS"]
    )
)

answer = pet_supplies["proper_noun_count"].mean()

print(answer)
```

```python
# Question 25

from nltk.tokenize import word_tokenize
from collections import Counter

all_tokens = []

for review in pet_supplies["review"]:
    tokens = word_tokenize(review, preserve_line=True)
    all_tokens.extend(tokens)

token_counts = Counter(all_tokens)

top_5_tokens = token_counts.most_common(5)

print(top_5_tokens)
```

```python
# Question 27
from textblob import TextBlob

pet_supplies["sentiment_score"] = pet_supplies["review"].apply(
    lambda x: TextBlob(x).sentiment.polarity
)

answer = pet_supplies.loc[
    pet_supplies["rating"] == 5,
    "sentiment_score"
].mean()

print(answer)
```
